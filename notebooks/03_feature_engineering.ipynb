```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.feature_selection import SelectKBest, f_classif
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Load the cleaned dataset
data = pd.read_csv('../data/processed/air_quality_cleaned.csv')

# Feature Creation: Generating a pollution index based on the concentration of key gases
def create_pollution_index(df):
    """Create a new feature 'pollution_index' as the average of key pollutants."""
    df['pollution_index'] = df[['CO(GT)', 'NOx(GT)', 'NO2(GT)']].mean(axis=1)
    return df

# Feature Scaling: Standardizing the features
def scale_features(df):
    """Scale numerical features using StandardScaler."""
    scaler = StandardScaler()
    scaled_data = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)
    return scaled_data

# Feature Selection: Using correlation matrix and ANOVA F-test
def feature_selection(df, target):
    """Selects the top K best features based on ANOVA F-test."""
    X = df.drop(columns=[target])
    y = df[target]
    selector = SelectKBest(f_classif, k=5)
    selector.fit(X, y)
    selected_features = X.columns[selector.get_support()]
    return df[selected_features]

# PCA for dimensionality reduction
def apply_pca(df, n_components=2):
    """Apply PCA to reduce the dimensionality of the dataset."""
    pca = PCA(n_components=n_components)
    pca_data = pca.fit_transform(df)
    return pd.DataFrame(pca_data, columns=[f'PC{i}' for i in range(1, n_components + 1)])

# Run Feature Engineering Pipeline
def feature_engineering_pipeline(filepath, target):
    df = pd.read_csv(filepath)
    df = create_pollution_index(df)
    df = scale_features(df)
    df_selected = feature_selection(df, target)
    pca_df = apply_pca(df_selected)
    return pca_df
```

```python
# Example of running the pipeline
if __name__ == "__main__":
    filepath = '../data/processed/air_quality_cleaned.csv'
    target = 'pollution_index'
    final_features = feature_engineering_pipeline(filepath, target)
    print(final_features.head())
```
